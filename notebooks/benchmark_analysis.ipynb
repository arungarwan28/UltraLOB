# UltraLOB — benchmark analysis

Reads the checked-in measurement artefact `reports/benchmark_results.json` and plots the
percentile profile and throughput per workload.

**Status:** provided for analysis; it has not been executed in this repository's build
environment (no Jupyter kernel was available). Run it with `jupyter lab` after
`bash scripts/run_benchmarks.sh` if you want to reproduce the figures.

All values come from the native harness on the machine that produced the report, not from the
browser build.

```python
import json
import pathlib

import matplotlib.pyplot as plt

path = pathlib.Path('reports/benchmark_results.json')
data = json.loads(path.read_text())
print('generated:', data['generated_at'])
print('cpu:', data['environment'].get('cpu'))
print('cores:', data['environment'].get('logical_cores'))
len(data['runs']), 'runs'
```

```python
labels = ['p50', 'p99', 'p99_9', 'p99_99', 'max']
keys = ['p50', 'p99', 'p99_9', 'p99_99', 'max']

for run in data['runs']:
    if run['repeat'] != 1:
        continue
    values = [run['latency_ns'][key] for key in keys]
    plt.figure(figsize=(6, 3))
    plt.bar(labels, values, color=['#35e0c8'] * 4 + ['#f5b64b'])
    plt.yscale('log')
    plt.ylabel('nanoseconds (log)')
    plt.title(f"{run['workload']} · {run['events']:,} events")
    plt.show()
```

```python
for row in data.get('summary', []):
    print(
        f"{row['workload']:>13} {row['events']:>10,} events  "
        f"{row['median_events_per_second']:>12,} msg/s  "
        f"p99 {row['median_p99_ns']:>6,} ns  "
        f"allocs {row['median_heap_allocations']}"
    )
```
